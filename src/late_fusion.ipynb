```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# Set random seed for reproducibility
np.random.seed(42)

# Parameters for the synthetic dataset
n_samples = 1000  # Number of samples per modality
n_features_wrist = 5  # Number of wrist features
n_features_chest = 5  # Number of chest features
n_classes = 3  # Number of classes

# Generate synthetic wrist data
X_wrist = np.random.rand(n_samples, n_features_wrist)  # Random features between 0 and 1
y_wrist = np.random.randint(0, n_classes, n_samples)  # Random class labels

# Generate synthetic chest data
X_chest = np.random.rand(n_samples, n_features_chest)  # Random features between 0 and 1
y_chest = y_wrist  # Same labels for wrist and chest data

# Split the data into training and testing sets
X_train_wrist, X_test_wrist, y_train_wrist, y_test_wrist = train_test_split(X_wrist, y_wrist, test_size=0.2, random_state=42)
X_train_chest, X_test_chest, y_train_chest, y_test_chest = train_test_split(X_chest, y_chest, test_size=0.2, random_state=42)

# Train separate models for wrist and chest data
model_wrist = RandomForestClassifier(n_estimators=50, random_state=42)
model_chest = RandomForestClassifier(n_estimators=50, random_state=42)

model_wrist.fit(X_train_wrist, y_train_wrist)
model_chest.fit(X_train_chest, y_train_chest)

# Predict probabilities for test sets
prob_wrist = model_wrist.predict_proba(X_test_wrist)
prob_chest = model_chest.predict_proba(X_test_chest)

# Perform late fusion by averaging the probabilities
prob_fused = (prob_wrist + prob_chest) / 2

# Final prediction by taking the class with the highest probability
y_pred_fused = np.argmax(prob_fused, axis=1)

# Evaluate the fused model
accuracy = accuracy_score(y_test_wrist, y_pred_fused)
report = classification_report(y_test_wrist, y_pred_fused)

print(f"Accuracy of the fused model: {accuracy:.2f}")
print("Classification Report of the fused model:")
print(report)
```
